```python
# ============================================================
# One-cell Google Colab script
# Military Aircraft Detection with YOLO11 (latest official)
# H100 + CUDA + checkpoints + TensorRT deployment
# ============================================================

# -----------------------------
# 1. Install dependencies
# -----------------------------
!pip install -q -U huggingface_hub ultralytics pyyaml onnx onnxsim numba

# Install TensorRT for Colab (CUDA 12.x based images)
!pip install -q tensorrt-cu12

# Optional, sometimes helpful for TRT export
!pip install -q onnx-graphsurgeon


# -----------------------------
# 2. Imports
# -----------------------------
import os
import random
import shutil
import time
import yaml

import numpy as np
import torch

from collections import defaultdict, Counter
from pathlib import Path
from PIL import Image
from IPython.display import display, Image as DisplayImage

from huggingface_hub import snapshot_download
from ultralytics import YOLO
from numba import cuda


# -----------------------------
# 3. CUDA/H100 configuration
# -----------------------------
if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA GPU not detected. Select an H100 runtime in "
        "Runtime > Change runtime type > GPU."
    )

CUDA_DEVICE_ID = 0
torch.cuda.set_device(CUDA_DEVICE_ID)

torch.backends.cudnn.benchmark = True
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True

try:
    torch.set_float32_matmul_precision("high")
except Exception:
    pass

DEVICE = CUDA_DEVICE_ID
GPU_NAME = torch.cuda.get_device_name(DEVICE)
CUDA_RUNTIME = torch.version.cuda
GPU_MEMORY_GB = (
    torch.cuda.get_device_properties(DEVICE).total_memory
    / 1024**3
)

print("CUDA available:", torch.cuda.is_available())
print("CUDA device:", DEVICE)
print("GPU:", GPU_NAME)
print("CUDA runtime:", CUDA_RUNTIME)
print("GPU memory:", round(GPU_MEMORY_GB, 2), "GB")


# -----------------------------
# 4. Configuration
# -----------------------------
REPO_ID = "a2015003713/military-aircraft-detection-dataset"

OUTPUT_ROOT = Path("/content/aircraft_yolo_dataset")
RUNS_ROOT = Path("/content/runs/detect")

MODEL_SIZE = "n"          # n, s, m, l, x
IMAGE_SIZE = 640
EPOCHS = 50

# -1 automatically selects a batch size based on GPU memory
BATCH_SIZE = -1

VALIDATION_RATIO = 0.20
RANDOM_SEED = 42

# Creates epoch10.pt, epoch20.pt, etc.
CHECKPOINT_PERIOD = 10

# Set True only to resume an interrupted full run
RESUME_TRAINING = False

# Run a short sanity check before full training
RUN_DEBUG_TRAINING = True

# Export ONNX in addition to TensorRT
EXPORT_ONNX_FALLBACK = True

FULL_RUN_NAME = f"military_aircraft_yolo11{MODEL_SIZE}"

# H100-friendly data loading
WORKERS = min(8, os.cpu_count() or 2)

random.seed(RANDOM_SEED)
torch.manual_seed(RANDOM_SEED)


# -----------------------------
# 5. CUDA detection-count kernel
# -----------------------------
@cuda.jit
def count_classes_kernel(class_ids, counts, number_of_classes):
    index = cuda.grid(1)

    if index < class_ids.size:
        class_id = class_ids[index]

        if 0 <= class_id < number_of_classes:
            cuda.atomic.add(counts, class_id, 1)


def count_detections_on_gpu(class_ids, number_of_classes):
    class_ids = np.asarray(class_ids, dtype=np.int32)

    if class_ids.size == 0:
        return np.zeros(number_of_classes, dtype=np.int32)

    gpu_class_ids = cuda.to_device(class_ids)

    gpu_counts = cuda.to_device(
        np.zeros(number_of_classes, dtype=np.int32)
    )

    threads_per_block = 256
    blocks_per_grid = (
        class_ids.size + threads_per_block - 1
    ) // threads_per_block

    count_classes_kernel[
        blocks_per_grid,
        threads_per_block
    ](
        gpu_class_ids,
        gpu_counts,
        number_of_classes
    )

    cuda.synchronize()

    return gpu_counts.copy_to_host()


# -----------------------------
# 6. Download dataset
# -----------------------------
print("\nDownloading dataset...")

hf_path = snapshot_download(
    repo_id=REPO_ID,
    repo_type="dataset",
    local_dir="/content/hf_aircraft_dataset"
)

dataset_path = Path(hf_path)

print("Dataset path:")
print(dataset_path)

print("\nDataset contents:")
for item in sorted(dataset_path.iterdir()):
    print(" -", item)


# -----------------------------
# 7. Locate images and labels
# -----------------------------
image_extensions = {
    ".jpg", ".jpeg", ".png", ".bmp",
    ".tif", ".tiff"
}

image_files = [
    file
    for file in dataset_path.rglob("*")
    if file.is_file()
    and file.suffix.lower() in image_extensions
]

preferred_label_dir = dataset_path / "annotations" / "yolo"

if preferred_label_dir.exists():
    label_files = list(preferred_label_dir.rglob("*.txt"))
else:
    label_files = list(dataset_path.rglob("*.txt"))

print("\nImages found:", len(image_files))
print("Labels found:", len(label_files))

if not image_files:
    raise FileNotFoundError(
        "No image files were found in the downloaded dataset."
    )

if not label_files:
    raise FileNotFoundError(
        "No YOLO annotation files were found."
    )


# -----------------------------
# 8. Match images and labels
# -----------------------------
images_by_stem = defaultdict(list)

for image_file in image_files:
    images_by_stem[image_file.stem].append(image_file)

matched_pairs = []
unmatched_labels = []
duplicate_stems = []

for label_file in label_files:
    candidates = images_by_stem.get(label_file.stem, [])

    if len(candidates) == 1:
        matched_pairs.append((candidates[0], label_file))
    elif len(candidates) == 0:
        unmatched_labels.append(label_file)
    else:
        duplicate_stems.append((label_file, candidates))

print("\nMatched pairs:", len(matched_pairs))
print("Unmatched labels:", len(unmatched_labels))
print("Duplicate stems:", len(duplicate_stems))

if duplicate_stems:
    print("\nDuplicate examples:")

    for label_file, candidates in duplicate_stems[:5]:
        print("Label:", label_file)
        print("Images:", candidates)

    raise RuntimeError(
        "Duplicate image stems found. "
        "Resolve ambiguous image-label pairs before training."
    )

if not matched_pairs:
    raise RuntimeError(
        "No matching image-label pairs were found."
    )


# -----------------------------
# 9. Validate images and labels
# -----------------------------
def validate_image(image_file):
    try:
        with Image.open(image_file) as image:
            image.verify()

        return True, ""

    except Exception as error:
        return False, str(error)


def validate_label(label_file):
    valid_rows = []
    errors = []

    content = label_file.read_text().strip()

    if not content:
        return valid_rows, ["empty label file"]

    for line_number, line in enumerate(
        content.splitlines(),
        start=1
    ):
        values = line.split()

        if len(values) != 5:
            errors.append(
                f"line {line_number}: expected 5 values, got {len(values)}"
            )
            continue

        try:
            class_id = int(values[0])
            coordinates = [
                float(value)
                for value in values[1:]
            ]

        except ValueError:
            errors.append(
                f"line {line_number}: non-numeric value"
            )
            continue

        if class_id < 0:
            errors.append(
                f"line {line_number}: negative class ID"
            )
            continue

        if not all(0 <= value <= 1 for value in coordinates):
            errors.append(
                f"line {line_number}: coordinate outside [0, 1]"
            )
            continue

        center_x, center_y, width, height = coordinates

        if width <= 0 or height <= 0:
            errors.append(
                f"line {line_number}: width or height is zero"
            )
            continue

        x_min = center_x - width / 2
        x_max = center_x + width / 2
        y_min = center_y - height / 2
        y_max = center_y + height / 2

        if not (
            0 <= x_min <= 1
            and 0 <= x_max <= 1
            and 0 <= y_min <= 1
            and 0 <= y_max <= 1
        ):
            errors.append(
                f"line {line_number}: box extends outside image"
            )
            continue

        valid_rows.append((class_id, coordinates))

    return valid_rows, errors


valid_pairs = []
invalid_images = []
invalid_labels = []
class_ids = set()

for image_file, label_file in matched_pairs:
    image_ok, image_error = validate_image(image_file)

    if not image_ok:
        invalid_images.append((image_file, image_error))
        continue

    rows, errors = validate_label(label_file)

    if not rows:
        invalid_labels.append((label_file, errors))
        continue

    for class_id, coordinates in rows:
        class_ids.add(class_id)

    valid_pairs.append((image_file, label_file))

print("\nValid pairs:", len(valid_pairs))
print("Invalid images:", len(invalid_images))
print("Invalid labels:", len(invalid_labels))
print("Class IDs:", sorted(class_ids))

if not valid_pairs:
    raise RuntimeError(
        "No valid image-label pairs remain."
    )

if not class_ids:
    raise RuntimeError(
        "No valid class IDs were found."
    )


# -----------------------------
# 10. Create YOLO directories
# -----------------------------
if OUTPUT_ROOT.exists():
    shutil.rmtree(OUTPUT_ROOT)

for split in ["train", "val"]:
    (OUTPUT_ROOT / "images" / split).mkdir(
        parents=True,
        exist_ok=True
    )

    (OUTPUT_ROOT / "labels" / split).mkdir(
        parents=True,
        exist_ok=True
    )


# -----------------------------
# 11. Split dataset
# -----------------------------
random.shuffle(valid_pairs)

validation_count = max(
    1,
    int(len(valid_pairs) * VALIDATION_RATIO)
)

validation_pairs = valid_pairs[:validation_count]
training_pairs = valid_pairs[validation_count:]

if not training_pairs:
    raise RuntimeError(
        "Training split is empty. Reduce VALIDATION_RATIO."
    )

print("\nTraining pairs:", len(training_pairs))
print("Validation pairs:", len(validation_pairs))


# -----------------------------
# 12. Copy files
# -----------------------------
def copy_pairs(pairs, split):
    used_names = set()

    for index, (image_file, label_file) in enumerate(pairs):
        base_name = image_file.stem

        if base_name in used_names:
            base_name = f"{base_name}_{index}"

        used_names.add(base_name)

        destination_image = (
            OUTPUT_ROOT
            / "images"
            / split
            / f"{base_name}{image_file.suffix.lower()}"
        )

        destination_label = (
            OUTPUT_ROOT
            / "labels"
            / split
            / f"{base_name}.txt"
        )

        shutil.copy2(image_file, destination_image)
        shutil.copy2(label_file, destination_label)


copy_pairs(training_pairs, "train")
copy_pairs(validation_pairs, "val")


# -----------------------------
# 13. Create data.yaml
# -----------------------------
number_of_classes = max(class_ids) + 1

class_names = [
    f"class_{index}"
    for index in range(number_of_classes)
]

missing_class_ids = (
    set(range(number_of_classes)) - class_ids
)

print("\nNumber of classes:", number_of_classes)
print("Class names:", class_names)
print("Missing class IDs:", sorted(missing_class_ids))

data_yaml_path = OUTPUT_ROOT / "data.yaml"

data_config = {
    "path": str(OUTPUT_ROOT.resolve()),
    "train": "images/train",
    "val": "images/val",
    "names": class_names
}

with open(data_yaml_path, "w") as file:
    yaml.safe_dump(
        data_config,
        file,
        sort_keys=False
    )

print("\nCreated data.yaml:")
print(data_yaml_path.read_text())


# -----------------------------
# 14. Verify image-label pairs
# -----------------------------
for split in ["train", "val"]:
    image_dir = OUTPUT_ROOT / "images" / split
    label_dir = OUTPUT_ROOT / "labels" / split

    image_stems = {
        file.stem
        for file in image_dir.iterdir()
        if file.is_file()
    }

    label_stems = {
        file.stem
        for file in label_dir.glob("*.txt")
    }

    print(f"\n{split.upper()} verification")
    print("Images:", len(image_stems))
    print("Labels:", len(label_stems))
    print("Images without labels:", len(image_stems - label_stems))
    print("Labels without images:", len(label_stems - image_stems))

    if image_stems != label_stems:
        raise RuntimeError(
            f"{split} images and labels do not match."
        )

for cache_file in OUTPUT_ROOT.rglob("*.cache"):
    cache_file.unlink()


# -----------------------------
# 15. Debug training (YOLO11)
# -----------------------------
if RUN_DEBUG_TRAINING:
    print("\nStarting 3-epoch debug training...")

    debug_model = YOLO("yolo11n.pt")

    debug_model.train(
        data=str(data_yaml_path),
        imgsz=IMAGE_SIZE,
        epochs=3,
        batch=4,
        device=DEVICE,
        workers=WORKERS,
        cache=False,
        pretrained=True,
        patience=2,
        amp=True,
        project=str(RUNS_ROOT),
        name="debug_aircraft",
        exist_ok=True,
        save=True,
        save_period=1
    )

    print("Debug training completed.")


# -----------------------------
# 16. Full H100 CUDA training (YOLO11)
# -----------------------------
print("\nStarting full H100 CUDA training...")

full_run_directory = RUNS_ROOT / FULL_RUN_NAME
last_checkpoint = full_run_directory / "weights" / "last.pt"

torch.cuda.empty_cache()
torch.cuda.reset_peak_memory_stats(DEVICE)
torch.cuda.synchronize()

training_start = time.perf_counter()

if RESUME_TRAINING:
    if not last_checkpoint.exists():
        raise FileNotFoundError(
            f"Resume checkpoint not found:\n{last_checkpoint}"
        )

    print("Resuming from:")
    print(last_checkpoint)

    model = YOLO(str(last_checkpoint))

    model.train(
        resume=True,
        device=DEVICE,
        workers=WORKERS,
        amp=True
    )

else:
    model = YOLO(f"yolo11{MODEL_SIZE}.pt")

    model.train(
        data=str(data_yaml_path),
        imgsz=IMAGE_SIZE,
        epochs=EPOCHS,
        batch=BATCH_SIZE,
        device=DEVICE,
        workers=WORKERS,
        amp=True,
        pretrained=True,
        patience=10,
        cache="ram",
        plots=True,

        # Checkpoints
        save=True,
        save_period=CHECKPOINT_PERIOD,

        project=str(RUNS_ROOT),
        name=FULL_RUN_NAME,
        exist_ok=True
    )

torch.cuda.synchronize()

training_seconds = time.perf_counter() - training_start
peak_memory_gb = (
    torch.cuda.max_memory_allocated(DEVICE)
    / 1024**3
)

print("\nCUDA training completed.")
print("Training time:", round(training_seconds / 60, 2), "minutes")
print(
    "Peak allocated GPU memory:",
    round(peak_memory_gb, 2),
    "GB"
)


# -----------------------------
# 17. Verify checkpoints
# -----------------------------
results_directory = RUNS_ROOT / FULL_RUN_NAME
weights_directory = results_directory / "weights"

best_checkpoint = weights_directory / "best.pt"
last_checkpoint = weights_directory / "last.pt"

print("\n" + "=" * 60)
print("CHECKPOINT SUMMARY")
print("=" * 60)
print("Run directory:", results_directory)
print("Best checkpoint:", best_checkpoint)
print("Last checkpoint:", last_checkpoint)

periodic_checkpoints = sorted(
    weights_directory.glob("epoch*.pt")
)

print("Periodic checkpoints:", len(periodic_checkpoints))

for checkpoint in periodic_checkpoints:
    print(" -", checkpoint.name)

if not best_checkpoint.exists():
    raise FileNotFoundError(
        f"Best checkpoint was not created:\n{best_checkpoint}"
    )

if not last_checkpoint.exists():
    raise FileNotFoundError(
        f"Last checkpoint was not created:\n{last_checkpoint}"
    )

print("Best checkpoint verified.")
print("Last checkpoint verified.")


# -----------------------------
# 18. Load best model
# -----------------------------
best_model = YOLO(str(best_checkpoint))

print("\nBest model loaded:")
print(best_checkpoint)


# -----------------------------
# 19. Display training plot
# -----------------------------
results_plot = results_directory / "results.png"

if results_plot.exists():
    display(
        DisplayImage(
            filename=str(results_plot),
            width=1000
        )
    )


# -----------------------------
# 20. CUDA validation
# -----------------------------
print("\nValidating best model on CUDA...")

torch.cuda.empty_cache()
torch.cuda.synchronize()

validation_results = best_model.val(
    data=str(data_yaml_path),
    split="val",
    imgsz=IMAGE_SIZE,
    batch=4,
    device=DEVICE,
    workers=WORKERS,
    amp=True,
    plots=True,
    save_json=True
)

torch.cuda.synchronize()

print("\nValidation metrics")
print("mAP50:", validation_results.box.map50)
print("mAP50-95:", validation_results.box.map)
print("Precision:", validation_results.box.mp)
print("Recall:", validation_results.box.mr)

if hasattr(validation_results, "results_dict"):
    print("\nFull validation results:")
    print(validation_results.results_dict)
else:
    print("\nFull validation results: (results_dict not available)")


# -----------------------------
# 21. Collect validation images
# -----------------------------
val_images = [
    file
    for file in (OUTPUT_ROOT / "images" / "val").iterdir()
    if file.suffix.lower() in image_extensions
]

random.shuffle(val_images)


# -----------------------------
# 22. Display predictions
# -----------------------------
print("\nDisplaying predictions...")

for image_file in val_images[:5]:
    predictions = best_model.predict(
        source=str(image_file),
        conf=0.25,
        imgsz=IMAGE_SIZE,
        device=DEVICE,
        half=True,
        verbose=False
    )

    annotated_image = predictions[0].plot()

    display(
        Image.fromarray(
            annotated_image[..., ::-1]
        )
    )


# -----------------------------
# 23. GPU-side detection counting
# -----------------------------
print("\nCounting detections with CUDA kernel...")

all_detected_class_ids = []

for image_file in val_images:
    predictions = best_model.predict(
        source=str(image_file),
        conf=0.25,
        imgsz=IMAGE_SIZE,
        device=DEVICE,
        half=True,
        verbose=False
    )

    result = predictions[0]

    if result.boxes is None or len(result.boxes) == 0:
        continue

    detected_ids = (
        result.boxes.cls
        .detach()
        .to("cuda")
        .to(torch.int32)
        .cpu()
        .numpy()
    )

    all_detected_class_ids.extend(
        detected_ids.tolist()
    )

number_of_classes = len(best_model.names)

gpu_counts = count_detections_on_gpu(
    all_detected_class_ids,
    number_of_classes
)

total_detections = int(gpu_counts.sum())

class_counts = Counter()

for class_id, count in enumerate(gpu_counts):
    if count > 0:
        class_name = best_model.names[class_id]
        class_counts[class_name] = int(count)

print("\nGPU-counted detections:", total_detections)
print("Detections by class:")

for class_name, count in sorted(class_counts.items()):
    print(f"{class_name}: {count}")


# -----------------------------
# 24. Benchmark PyTorch CUDA inference
# -----------------------------
if val_images:
    print("\nBenchmarking PyTorch CUDA inference...")

    benchmark_image = str(val_images[0])

    for _ in range(5):
        best_model.predict(
            source=benchmark_image,
            imgsz=IMAGE_SIZE,
            device=DEVICE,
            half=True,
            verbose=False
        )

    torch.cuda.synchronize()

    benchmark_start = time.perf_counter()

    for _ in range(20):
        best_model.predict(
            source=benchmark_image,
            imgsz=IMAGE_SIZE,
            device=DEVICE,
            half=True,
            verbose=False
        )

    torch.cuda.synchronize()

    benchmark_seconds = (
        time.perf_counter() - benchmark_start
    )

    average_ms = benchmark_seconds / 20 * 1000
    approximate_fps = 1000 / average_ms

    print("Average inference:", round(average_ms, 2), "ms")
    print("Approximate FPS:", round(approximate_fps, 2))


# -----------------------------
# 25. Export TensorRT FP16 (SAFE)
# -----------------------------
print("\nExporting TensorRT FP16 engine...")

engine_path = None

try:
    import tensorrt  # noqa: F401
except ImportError:
    print("WARNING: tensorrt Python package not available. Skipping TensorRT export.")
    print("If you need TensorRT, ensure tensorrt-cu12 (or matching CUDA version) is installed.")
else:
    try:
        engine_path = best_model.export(
            format="engine",
            imgsz=IMAGE_SIZE,
            half=True,
            dynamic=False,
            batch=1,
            workspace=16,
            device=DEVICE
        )
        print("TensorRT engine:")
        print(engine_path)
    except Exception as e:
        print("WARNING: TensorRT export failed with error:")
        print(e)
        print("Continuing without TensorRT engine.")


# -----------------------------
# 26. Test TensorRT engine (if available)
# -----------------------------
engine_model = None

if engine_path is not None and Path(engine_path).exists():
    print("\nTesting TensorRT engine...")

    engine_model = YOLO(str(engine_path))

    if val_images:
        test_image = str(val_images[0])

        engine_predictions = engine_model.predict(
            source=test_image,
            imgsz=IMAGE_SIZE,
            conf=0.25,
            device=DEVICE,
            batch=1,
            verbose=False
        )

        engine_annotated = engine_predictions[0].plot()

        display(
            Image.fromarray(
                engine_annotated[..., ::-1]
            )
        )
else:
    print("\nNo valid TensorRT engine to test (skipping).")


# -----------------------------
# 27. Benchmark TensorRT engine (if available)
# -----------------------------
if engine_model is not None and val_images:
    print("\nBenchmarking TensorRT inference...")

    test_image = str(val_images[0])

    for _ in range(5):
        engine_model.predict(
            source=test_image,
            imgsz=IMAGE_SIZE,
            device=DEVICE,
            batch=1,
            verbose=False
        )

    torch.cuda.synchronize()

    trt_start = time.perf_counter()

    for _ in range(20):
        engine_model.predict(
            source=test_image,
            imgsz=IMAGE_SIZE,
            device=DEVICE,
            batch=1,
            verbose=False
        )

    torch.cuda.synchronize()

    trt_seconds = time.perf_counter() - trt_start
    trt_average_ms = trt_seconds / 20 * 1000
    trt_fps = 1000 / trt_average_ms

    print("TensorRT average inference:", round(trt_average_ms, 2), "ms")
    print("TensorRT approximate FPS:", round(trt_fps, 2))
else:
    print("\nNo TensorRT engine to benchmark (skipping).")


# -----------------------------
# 28. Export ONNX fallback (COMPATIBLE)
# -----------------------------
if EXPORT_ONNX_FALLBACK:
    print("\nExporting ONNX fallback...")

    onnx_path = best_model.export(
        format="onnx",
        imgsz=IMAGE_SIZE,
        half=False,
        dynamic=False,
        opset=17
        # 'simplify' removed for compatibility across ultralytics versions
    )

    print("ONNX model:")
    print(onnx_path)
else:
    onnx_path = None


# -----------------------------
# 29. Final summary
# -----------------------------
print("\n" + "=" * 60)
print("PROCESS COMPLETED SUCCESSFULLY")
print("=" * 60)
print("GPU:", GPU_NAME)
print("Best PyTorch checkpoint:", best_checkpoint)
print("Last PyTorch checkpoint:", last_checkpoint)
print("Dataset YAML:", data_yaml_path)
print("TensorRT engine:", engine_path if engine_path else "(not created)")
print("ONNX fallback:", onnx_path if EXPORT_ONNX_FALLBACK else "(disabled)")
print("Training directory:", results_directory)
```
